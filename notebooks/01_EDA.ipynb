```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

print("All libraries imported successfully!")
```

```python
# Load dataset
file_path = r"C:\Users\E106006\OneDrive - Road Accident Fund\Desktop\github Portfolio\Credit Risk\Data\credit_data.csv"
df = pd.read_csv(file_path)

# Remove index column if it exists
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

# Create Risk column based on financial rules
def assign_risk(row):
    risk_score = 0

    # Credit amount risk
    if row['Credit amount'] > 10000:
        risk_score += 3
    elif row['Credit amount'] > 5000:
        risk_score += 2
    elif row['Credit amount'] > 2000:
        risk_score += 1

    # Duration risk
    if row['Duration'] > 36:
        risk_score += 3
    elif row['Duration'] > 24:
        risk_score += 2
    elif row['Duration'] > 12:
        risk_score += 1

    # Age risk
    if row['Age'] < 25:
        risk_score += 2
    elif row['Age'] < 30:
        risk_score += 1

    # Saving accounts risk
    if pd.isna(row['Saving accounts']) or row['Saving accounts'] == 'little':
        risk_score += 2
    elif row['Saving accounts'] == 'moderate':
        risk_score += 1

    # Checking account risk
    if pd.isna(row['Checking account']) or row['Checking account'] == 'little':
        risk_score += 2
    elif row['Checking account'] == 'moderate':
        risk_score += 1

    # Final classification
    return 'bad' if risk_score >= 6 else 'good'

# Add Risk column
df['Risk'] = df.apply(assign_risk, axis=1)

# Risk distribution
risk_counts = df['Risk'].value_counts()

good_count = risk_counts.get('good', 0)
bad_count = risk_counts.get('bad', 0)

print("Risk Distribution:")
print(risk_counts)

if bad_count > 0:
    print(f"\nImbalance Ratio (Good:Bad): {good_count / bad_count:.2f}:1")
else:
    print("\nNo 'bad' records found.")

# Save updated dataset
df.to_csv(file_path, index=False)

print("\nDataset saved successfully with the Risk column.")

# Preview
print(df.head())
```
